### Data Preprocessing

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
# Extract Hour from the raw Time string instead of one-hot encoding every minute value
# (Time has ~1439 unique values -- one-hot encoding it directly caused the MemoryError)
df['Hour'] = pd.to_datetime(df['Time'], format='%H:%M', errors='coerce').dt.hour
df['Hour'] = df['Hour'].fillna(df['Hour'].median())

# Drop target, IDs, near-empty column, coordinates, and very high-cardinality columns
# (Local_Authority_(District): 422 categories, Police_Force: 51 categories, raw Time: ~1439 categories)
X = df.drop(columns=[
    'Accident_Severity', 'Accident_Index', 'Accident Date', 'Carriageway_Hazards',
    'Latitude', 'Longitude', 'Time', 'Local_Authority_(District)', 'Police_Force'
])
y = df['Accident_Severity']
```

```python
X.columns
```

```python
num_cols = X.select_dtypes(include=[int, float]).columns.tolist()
cat_cols = X.select_dtypes(include=[str]).columns.tolist()
```

```python
df['Month'].unique()
```

```python
X.columns
```

```python
for col in X.columns:
    print(col, len(df[col].unique()))
```

```python
# Nominal columns (Time, Local_Authority_(District), Police_Force removed -- see fix above;
# Time is now the numeric 'Hour' column, and the other two were dropped for high cardinality)
nominal_cols = [
    'Day_of_Week',
    'Junction_Control',
    'Junction_Detail',
    'Urban_or_Rural_Area',
    'Vehicle_Type'
]

# Ordinal columns
ordinal_cols = [
    'Month',
    'Weather_Conditions',
    'Road_Surface_Conditions',
    'Road_Type',
    'Light_Conditions'
]

# NOTE: categories_order must have exactly one list per entry in ordinal_cols (5 lists).
# The original had a 6th list intended for 'Carriageway_Hazards', which is dropped from X --
# that extra list has been removed since it would raise a shape-mismatch error in OrdinalEncoder.
categories_order = [
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    [
        'Fine no high winds',
        'Other',
        'Raining no high winds',
        'Fine + high winds',
        'Raining + high winds',
        'Snowing no high winds',
        'Snowing + high winds',
        'Fog or mist'
    ],
    [
        'Dry',
        'Wet or damp',
        'Snow',
        'Frost or ice',
        'Flood over 3cm. deep'
    ],
    [
        'One way street',
        'Single carriageway',
        'Dual carriageway',
        'Slip road',
        'Roundabout',
        'Unknown'
    ],
    [
        'Daylight',
        'Darkness - lights lit',
        'Darkness - lights unlit',
        'Darkness - no lighting',
        'Darkness - lighting unknown'
    ]
]
```

```python
# Split BEFORE fitting any preprocessing -- fitting the scaler/encoder on data that includes
# the test set would leak test information into preprocessing (see earlier leakage discussion).
# stratify=y is used because the target is imbalanced, so the split keeps class proportions.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=23, stratify=y
)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
nominal_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoding', OneHotEncoder(sparse_output=False, handle_unknown='ignore', drop='first'))
])
```

```python
ordinal_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(categories=categories_order))
])
```

```python
numerical_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('Scaling', StandardScaler())
])
```

```python
preprocessor = ColumnTransformer([
    ('Nominal', nominal_pipe, nominal_cols),
    ('Ordinal', ordinal_pipe, ordinal_cols),
    ('Numerical', numerical_pipe, num_cols)
])
preprocessor
```

```python
# Fit ONLY on training data, then transform both train and test with the same fitted preprocessor
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

```python
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(y_train)
y_test = label_encoder.transform(y_test)
label_encoder.classes_
```

```python
df['Accident_Severity'].value_counts()
```

```python
target_classes = df['Accident_Severity'].nunique()
target_classes
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=23)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)
X_test_resampled, y_test_resampled = smote.fit_resample(X_test, y_test)
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

balanced_weights = compute_class_weight(
    class_weight='balanced', classes=np.unique(y_train), y=y_train_resampled
)
softened_weights = np.sqrt(balanced_weights)
class_weight_dict = dict(zip(np.unique(y_train_resampled), softened_weights))
class_weight_dict
```

```python
X_train_resampled.shape, X_test_resampled.shape, y_train_resampled.shape, y_test_resampled.shape
```

### Model Building & Evaluation

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

import warnings
warnings.filterwarnings('ignore')
```

```python
model = Sequential()

model.add(Input(shape=(X_train.shape[1],))) # Input layer

model.add(Dense(units=64, activation='relu')) # Hidden layer
model.add(Dense(units=32, activation='relu')) # Hidden layer

model.add(Dense(units=target_classes, activation='softmax'))
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(
    X_train_resampled, y_train_resampled,
    epochs=20,
    validation_data=(X_test_resampled, y_test_resampled)
)
```

```python
history.history.keys()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(121)
sns.lineplot(x=history.epoch, y=history.history['accuracy'], label='train accuracy')
sns.lineplot(x=history.epoch, y=history.history['val_accuracy'], label='validation accuracy')
plt.title('Train Accuracy vs Validation Accuracy')

plt.subplot(122)
sns.lineplot(x=history.epoch, y=history.history['loss'], label='train loss')
sns.lineplot(x=history.epoch, y=history.history['val_loss'], label='validation loss')
plt.title('Train Loss vs Validation Loss')

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np

y_pred_probs = model.predict(X_test_resampled)
y_pred = np.argmax(y_pred_probs, axis=1)

print(classification_report(y_test_resampled, y_pred, target_names=label_encoder.classes_))
print(confusion_matrix(y_test_resampled, y_pred))
```

### Hyperparameter tuning

```python
import optuna

from keras.regularizers import L1L2
from keras.layers import Dropout, BatchNormalization
from keras.optimizers import RMSprop, SGD, Adam
from keras.callbacks import EarlyStopping, ModelCheckpoint

from sklearn.metrics import f1_score
```

```python
X_val = X_test_resampled[:500]
y_val = y_test_resampled[:500]

def Objective(trial):
    n_layers = trial.suggest_int('n_layers', 1, 4)
    epochs = trial.suggest_int('epochs', 10, 40)

    batch_size = trial.suggest_int('batch_size', 16, 256)
    lr_rate = trial.suggest_float('lr_rate', 1e-4, 1e-2, log=True)
    activation = trial.suggest_categorical('activation', ['relu'])
    optimizer = trial.suggest_categorical('optimizer', ['Adam', 'RMSprop'])

    model = Sequential()
    model.add(Input(shape=(X_train.shape[1],)))

    for i in range(n_layers):
        units = trial.suggest_int(f'unit{i}', 8, 200)
        dropout = trial.suggest_float(f'dropout{i}', 0.1, 0.5)
        reg_strength = trial.suggest_float(f'reg_strength{i}', 1e-5, 1e-3, log=True)

        model.add(Dense(units=units, activation=activation,
                         kernel_regularizer=L1L2(l1=reg_strength, l2=reg_strength)))
        model.add(BatchNormalization())
        model.add(Dropout(rate=dropout))

    model.add(Dense(units=target_classes, activation='softmax'))

    early_stopping = EarlyStopping(monitor='val_loss', min_delta=0.001, patience=3, restore_best_weights=True)
    checkpoint = ModelCheckpoint(filepath="best_model.keras", monitor='val_loss',
                                  save_best_only=True, save_weights_only=False, mode='min')

    if optimizer == 'Adam':
        opt = Adam(learning_rate=lr_rate)
    elif optimizer == 'RMSprop':
        opt = RMSprop(learning_rate=lr_rate)
    else:
        opt = SGD(learning_rate=lr_rate)

    model.compile(optimizer=opt, loss='sparse_categorical_crossentropy', metrics=['accuracy'])

    history = model.fit(
        X_train_resampled, y_train_resampled,
        epochs=epochs,
        validation_data=(X_val, y_val),
        batch_size=batch_size,
        class_weight=class_weight_dict,
        callbacks=[early_stopping, checkpoint]
    )

    y_pred_prob = model.predict(X_val, verbose=0)
    y_val_pred = np.argmax(y_pred_prob, axis=1)

    return f1_score(y_val, y_val_pred, average="macro")
```

```python
study = optuna.create_study(direction='maximize')
study.optimize(Objective, n_trials=10)
```

```python
study.best_params
```

```python
best_params = study.best_params

n_layers = best_params['n_layers']
epochs = best_params['epochs']
lr_rate = best_params['lr_rate']
batch_size = best_params['batch_size']
activation = best_params['activation']
optimizers = best_params['optimizer']

model = Sequential()

model.add(Input(shape=(X_train.shape[1],)))

for i in range(n_layers):
    units = best_params[f'unit{i}']
    dropout = best_params[f'dropout{i}']
    strength = best_params[f'reg_strength{i}']

    model.add(Dense(units=units, activation=activation, kernel_regularizer=L1L2(l1=strength, l2=strength)))
    model.add(Dropout(dropout))
    model.add(BatchNormalization())

model.add(Dense(units=target_classes, activation='softmax'))

early_stopping = EarlyStopping(monitor='val_accuracy', min_delta=0.01, patience=3, restore_best_weights=True)

model.compile(optimizer=Adam(learning_rate=lr_rate), loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history = model.fit(
    X_train_resampled, y_train_resampled,
    epochs=epochs,
    validation_data=(X_test_resampled, y_test_resampled),
    batch_size=batch_size,
    class_weight=class_weight_dict,
    callbacks=[early_stopping]
)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(121)
sns.lineplot(x=history.epoch, y=history.history['accuracy'], label='train accuracy')
sns.lineplot(x=history.epoch, y=history.history['val_accuracy'], label='validation accuracy')
plt.title('Train Accuracy vs Validation Accuracy')

plt.subplot(122)
sns.lineplot(x=history.epoch, y=history.history['loss'], label='train loss')
sns.lineplot(x=history.epoch, y=history.history['val_loss'], label='validation loss')
plt.title('Train Loss vs Validation Loss')

plt.tight_layout()
plt.show()
```
